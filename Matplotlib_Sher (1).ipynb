# Saving Plots

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([6, 3, 1, 4, 99])

fig = plt.figure()

ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

ax.plot(x, y)

plt.savefig("basicplot.png")

plt.show()
```
